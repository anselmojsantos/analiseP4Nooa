```python
# Classificacao por zona usando zonas_sp.geojson (Centro, Norte, Sul, Leste, Oeste)
import geopandas as gpd

zonas = gpd.read_file(os.path.join(RAIZ, "zonas_sp.geojson"))

# Expande as zonas em 100 m (EPSG:31983) para incluir Cidade Tiradentes 2 (~34 m fora da Leste)
zonas["geometry"] = zonas.to_crs("EPSG:31983").geometry.buffer(40).to_crs("EPSG:4326").geometry

gdf = gpd.GeoDataFrame(
    df_sp,
    geometry=gpd.points_from_xy(df_sp["longitude"], df_sp["latitude"]),
    crs="EPSG:4326",
)
gdf = gpd.sjoin(gdf, zonas, how="left", predicate="within")
gdf["zona"] = gdf["zona"].fillna("Fora de SP")

df_zona = gdf.drop(columns=["geometry", "index_right"]).copy()
print(df_zona["zona"].value_counts())
```

```python
df_cemaden_station = df_zona[["id_estacao","station_code","ibge_code", "name", "city","latitude","longitude", "zona",
                             "network_id", "network_acronym","station_type_id","station_type_description","installation_date",
                             "registration_date","last_transmission","inactive_since","collected_at"]].copy()
```

```python
df_cemaden_station['installation_date'] = pd.to_datetime(df_cemaden_station['installation_date']).dt.date
df_cemaden_station['registration_date'] = pd.to_datetime(df_cemaden_station['registration_date']).dt.date
df_cemaden_station['last_transmission'] = pd.to_datetime(df_cemaden_station['last_transmission']).dt.date
df_cemaden_station['inactive_since'] = pd.to_datetime(df_cemaden_station['inactive_since']).dt.date
df_cemaden_station['collected_at'] = pd.to_datetime(df_cemaden_station['collected_at']).dt.date

df_cemaden_station.head()
```

```python
# Mapa das estacoes com pinos coloridos por zona (Icon)
import folium

cores = {
    "Centro": "gray",
    "Norte": "darkblue",
    "Sul": "darkgreen",
    "Leste": "red",
    "Oeste": "darkred",
    "Fora de SP": "lightgray",
}

m = folium.Map(
    location=[df_sp["latitude"].mean(), df_sp["longitude"].mean()],
    zoom_start=12,
)

folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=m&x={x}&y={y}&z={z}",
    attr="Google",
    name="Google Maps",
).add_to(m)

for _, r in df_cemaden_station.iterrows():
    cor = cores.get(r["zona"], "gray")
    folium.Marker(
        location=[r["latitude"], r["longitude"]],
        popup=f"{r['name']} - {r['zona']}",
        tooltip=r["name"],
        icon=folium.Icon(color=cor),
    ).add_to(m)

CAMINHO = os.path.join(RAIZ, "cemaden_stations_map.html")
m.save(CAMINHO)
print("Mapa salvo em:", CAMINHO)
m
```

```python
df_station = df_cemaden_station.copy()

df_station['installation_date'] = pd.to_datetime(df_station['installation_date']).dt.normalize()
df_station['last_transmission'] = pd.to_datetime(df_station['last_transmission']).dt.normalize()
df_station['inactive_since'] = pd.to_datetime(df_station['inactive_since']).dt.normalize()

today = pd.Timestamp.now().normalize()

df_station['dias_inativa'] = (today - df_station['last_transmission']).dt.days
df_station['anos_inativa'] = (df_station['dias_inativa'] / 365.25).round(1)

df_station.head(10)

```

```python
df_station['dias_entre_inativacao_e_transmissao'] = (
    df_station['last_transmission'] - df_station['inactive_since']
).dt.days

df_station[['name', 'zona', 'inactive_since', 'last_transmission',
            'dias_entre_inativacao_e_transmissao']].head(5)
```

```python
group_zona = (
    df_cemaden_station
    .groupby('zona')
    .agg(
        qtd_estacoes=('id_estacao', 'count'),
        ativas=('status', lambda x: (x == 'ativo').sum()),
        inativas=('status', lambda x: (x == 'inativa').sum()),
    )
    .sort_values('qtd_estacoes', ascending=False)
)

group_zona['pct_ativas'] = (group_zona['ativas'] / group_zona['qtd_estacoes'] * 100).round(1)
group_zona['pct_inativas'] = (group_zona['inativas'] / group_zona['qtd_estacoes'] * 100).round(1)

group_zona
```

```python
df_reading = pd.read_sql("SELECT * FROM cemaden_reading", engine)
df_sensor  = pd.read_sql("SELECT * FROM cemaden_sensor", engine)

print(f"Leituras:  {len(df_reading)}")
print(f"Estações:  {len(df_station)}")
print(f"Sensores:  {len(df_sensor)}")
```

```python
df_reading["observation_date"] = pd.to_datetime(df_reading["observation_date"])

print("Período das leituras:")
print(f"  De:  {df_reading['observation_date'].min()}")
print(f"  Até: {df_reading['observation_date'].max()}")
print(f"  Dias distintos: {df_reading['observation_date'].dt.date.nunique()}")
```

```python
# Contar leituras por sensor_id
leituras_por_sensor = (
    df_reading
    .groupby("sensor_id")
    .size()
    .reset_index(name="qtd_leituras")
    .sort_values("qtd_leituras", ascending=False)
)

# Juntar com a descrição do sensor
leituras_por_sensor = leituras_por_sensor.merge(
    df_sensor[["id", "description"]],
    left_on="sensor_id",
    right_on="id",
    how="left"
)

leituras_por_sensor[["sensor_id", "description", "qtd_leituras"]]
```

```python
# 1. Juntar leitura com estação (pelo station_code)
# 2. Juntar com sensor (pelo sensor_id)

df_full = (
    df_reading
    .merge(
        df_station[["station_code", "name", "city", "zona",
                    "latitude", "longitude"]],
        on="station_code",
        how="left"
    )
    .merge(
        df_sensor[["id", "description"]],
        left_on="sensor_id",
        right_on="id",
        how="left",
        suffixes=("", "_sensor")
    )
)

print(f"Total de linhas após merge: {len(df_full)}")
print(f"Colunas: {df_full.columns.tolist()}")
print()
df_full.head()
```
