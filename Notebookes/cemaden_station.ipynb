```python
new_df = df_sp.drop(columns=['cota_alerta', 'cota_atencao', 'cota_transbordamento', 'raw_payload']).reset_index(drop=True)  # Drop unnecessary columns
```

```python
# Classificacao por zona usando zonas_sp.geojson (Centro, Norte, Sul, Leste, Oeste)
import geopandas as gpd

zonas = gpd.read_file(os.path.join(RAIZ, "zonas_sp.geojson"))

# Expande as zonas em 100 m (EPSG:31983) para incluir Cidade Tiradentes 2 (~34 m fora da Leste)
zonas["geometry"] = zonas.to_crs("EPSG:31983").geometry.buffer(100).to_crs("EPSG:4326").geometry

gdf = gpd.GeoDataFrame(
    new_df,
    geometry=gpd.points_from_xy(new_df["longitude"], new_df["latitude"]),
    crs="EPSG:4326",
)
gdf = gpd.sjoin(gdf, zonas, how="left", predicate="within")
gdf["zona"] = gdf["zona"].fillna("Fora de SP")

new_df = gdf.drop(columns=["geometry", "index_right"]).copy()
print(new_df["zona"].value_counts())
```

```python
# DataFrame resumido das estacoes ja classificadas por zona
df_cemaden_station = new_df[["id_estacao","station_code","ibge_code", "name", "city","latitude","longitude", "zona",
                             "network_id", "network_acronym","station_type_id","station_type_description","installation_date",
                             "registration_date","inactive_since","collected_at"]].copy()
print(df_cemaden_station.to_string(index=False))
```

```python
# Mapa das estacoes com pinos coloridos por zona (Icon)
import folium

cores = {
    "Centro": "gray",
    "Norte": "darkblue",
    "Sul": "darkgreen",
    "Leste": "red",
    "Oeste": "darkred",
    "Fora de SP": "lightgray",
}

m = folium.Map(
    location=[new_df["latitude"].mean(), new_df["longitude"].mean()],
    zoom_start=12,
)

folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=m&x={x}&y={y}&z={z}",
    attr="Google",
    name="Google Maps",
).add_to(m)

for _, r in new_df.iterrows():
    cor = cores.get(r["zona"], "gray")
    folium.Marker(
        location=[r["latitude"], r["longitude"]],
        popup=f"{r['name']} - {r['zona']}",
        tooltip=r["name"],
        icon=folium.Icon(color=cor),
    ).add_to(m)

CAMINHO = os.path.join(RAIZ, "cemaden_stations_map.html")
m.save(CAMINHO)
print("Mapa salvo em:", CAMINHO)

# Exibe o mapa dentro do VSCode via iframe Base64 (contorna o aviso de "confiança" do folium/branca)
import base64
from IPython.display import display, IFrame

b64 = base64.b64encode(m._repr_html_().encode("utf-8")).decode("ascii")
display(IFrame(f"data:text/html;base64,{b64}", width="100%", height=600))
```
